   (b) Can a field line of an electrostatic field close on itself? Use $\oint\vec{E}\cdot d\vec{l}$ to argue.

:::{dropdown} Answers

   (b) $\sigma/\epsilon_0$ between the planes, pointing from $+\sigma$ to $-\sigma$, and $0$ outside.

   (c) No Gaussian surface has $\vec{E}$ constant in magnitude and simple in direction over it, so the flux integral can't be pulled apart to solve for $E$.

4. (a) $\rho = \epsilon_0\nabla\cdot\vec{E} = \dfrac{\epsilon_0}{r^2}\dfrac{d}{dr}(kr^5) = 5\epsilon_0kr^2$. Both ways give $Q = 4\pi\epsilon_0kR^5$.

   (b) $\nabla\times\vec{E}_1 = \vec{0}$, so $\vec{E}_1$ can be electrostatic. $\nabla\times\vec{E}_2 = 2k\hat{z} \ne \vec{0}$, so $\vec{E}_2$ cannot.

   (c) Between the charges, $1/x^2 = 4/(3 - x)^2$ gives $x = 1$.

## Coulomb's Law and Superposition

A point charge $q'$ at $\vec{r}'$ exerts on a charge $q$ at $\vec{r}$ the force

$$
\vec{F} = \frac{1}{4\pi\epsilon_0}\frac{qq'}{\mathscr{r}^2}\,\hat{\mathscr{r}}, \qquad \vec{\mathscr{r}} = \vec{r} - \vec{r}' ,
$$

with $\epsilon_0 = 8.854 \times 10^{-12}\ \text{C}^2/(\text{N m}^2)$. Forces from many charges **add** (superposition), which is an experimental fact, not a logical necessity. Dividing out the test charge defines the **electric field** $\vec{E} = \vec{F}/q$, a property of the sources alone.

For a continuous distribution, sum over charge elements $dq = \rho\,d\tau'$ (or $\lambda\,dl'$, $\sigma\,da'$):

$$
\vec{E}(\vec{r}) = \frac{1}{4\pi\epsilon_0}\int\frac{\rho(\vec{r}')}{\mathscr{r}^2}\,\hat{\mathscr{r}}\,d\tau' .
$$ (eq-coulomb-integral)

This is a vector integral: each element contributes in a different direction, so integrate components, and use symmetry to see which components cancel before computing anything.

::::{prf:example} A charged ring, then a disk
:label: ex-ring-disk

Find $\vec{E}$ on the axis of a ring of radius $a$ carrying total charge $Q$, a distance $z$ from its center. Then use it to find the field on the axis of a uniformly charged disk.

:::{dropdown} Solution
Every element is at distance $\mathscr{r} = \sqrt{a^2 + z^2}$. The components perpendicular to the axis cancel in opposite pairs. The axial component of each element's field is $\cos\alpha = z/\mathscr{r}$ times its magnitude, so

$$
E_z = \frac{1}{4\pi\epsilon_0}\frac{Qz}{(a^2 + z^2)^{3/2}} .
$$

Far away ($z\gg a$) this is $Q/(4\pi\epsilon_0z^2)$, a point charge, as it should be.

A disk of surface density $\sigma$ is a set of rings of radius $s$ and charge $\sigma\,2\pi s\,ds$:

$$
E_z = \frac{\sigma z}{2\epsilon_0}\int_0^R\frac{s\,ds}{(s^2 + z^2)^{3/2}} = \frac{\sigma}{2\epsilon_0}\left(1 - \frac{z}{\sqrt{R^2 + z^2}}\right) .
$$

As $R\to\infty$ this becomes $\sigma/(2\epsilon_0)$, independent of distance: the field of an infinite charged plane.
:::
::::

```python
# The disk integral, symbolically, and its two limits
s, z, R, sig, e0 = sp.symbols("s z R sigma epsilon_0", positive=True)
Ez = sig * z / (2 * e0) * sp.integrate(s / (s**2 + z**2)**sp.Rational(3, 2), (s, 0, R))
Ez = sp.simplify(Ez)
display(Math(r"E_z^{\text{disk}} = " + sp.latex(Ez)))
display(Math(r"\lim_{R\to\infty} E_z = " + sp.latex(sp.limit(Ez, R, sp.oo))))
display(Math(r"\lim_{z\to\infty} z^2 E_z = " + sp.latex(sp.limit(z**2 * Ez, z, sp.oo)) + r" = \frac{Q}{4\pi\epsilon_0}\ \text{ with } Q = \sigma\pi R^2"))
```

```python
#| label: fig-ring-field
#| caption: Field lines of a charged ring (seen edge-on, in the xz-plane) computed by direct numerical integration of eq-coulomb-integral. Near the ring the field is complicated; far away it approaches the radial field of a point charge.
a = 1.0
phi = np.linspace(0, 2 * np.pi, 200, endpoint=False)
src = np.c_[a * np.cos(phi), a * np.sin(phi), np.zeros_like(phi)]     # ring elements, equal charges
x = np.linspace(-3, 3, 61); zg = np.linspace(-3, 3, 61)
X, Z = np.meshgrid(x, zg)
Ex = np.zeros_like(X); Ezg = np.zeros_like(X)
for sx, sy, sz in src:
    dx, dy, dz = X - sx, -sy, Z - sz
    d3 = (dx**2 + dy**2 + dz**2)**1.5 + 1e-9
    Ex += dx / d3; Ezg += dz / d3
fig, ax = plt.subplots(figsize=(6, 6))
ax.streamplot(X, Z, Ex, Ezg, density=1.4, color="tab:blue", linewidth=0.9)
ax.plot([-a, a], [0, 0], "ro", ms=8)
ax.set_aspect("equal"); ax.set_xlabel("$x/a$"); ax.set_ylabel("$z/a$"); plt.show()
```

## Gauss's Law

The flux of a point charge's field through any closed surface around it is $q/\epsilon_0$, whatever the surface's shape (E-M1: the divergence of $\hat{r}/r^2$ is $4\pi\delta^3$). By superposition, for any charge distribution

$$
\oint_S\vec{E}\cdot d\vec{a} = \frac{Q_{\text{enc}}}{\epsilon_0}, \qquad\text{equivalently}\qquad \nabla\cdot\vec{E} = \frac{\rho}{\epsilon_0} .
$$ (eq-gauss)

The differential form follows from the integral form by the divergence theorem, and also directly from {eq}`eq-coulomb-integral` using $\nabla\cdot(\hat{\mathscr{r}}/\mathscr{r}^2) = 4\pi\delta^3(\vec{\mathscr{r}})$.

The other half of electrostatics is that the field is **curl-free**:

$$
\nabla\times\vec{E} = \vec{0} ,
$$

because each point charge's field is radial and spherically symmetric, and a sum of curl-free fields is curl-free. Together, divergence and curl determine $\vec{E}$ (Helmholtz theorem).

:::{important} When Gauss's law is a shortcut
{eq}`eq-gauss` is always true, but it **solves** for $\vec{E}$ only when symmetry makes $\vec{E}$ constant in magnitude and simple in direction over a well-chosen surface. Three symmetries qualify: spherical (use a sphere), cylindrical (a coaxial cylinder), and planar (a pillbox).
:::

::::{prf:example} A cylinder with nonuniform charge
:label: ex-cylinder-gauss

A long cylinder of radius $R$ has charge density $\rho = \rho_0s/R$ (growing linearly from the axis). Find $\vec{E}$ everywhere.

:::{dropdown} Solution
By symmetry $\vec{E} = E(s)\,\hat{s}$. Use a coaxial Gaussian cylinder of radius $s$ and length $L$. Its flux is $E\cdot2\pi sL$.

Inside ($s < R$): $Q_{\text{enc}} = \int_0^s\rho_0\frac{s'}{R}2\pi s'L\,ds' = \frac{2\pi\rho_0Ls^3}{3R}$, so $E = \dfrac{\rho_0s^2}{3\epsilon_0R}$.

Outside ($s > R$): $Q_{\text{enc}} = \frac{2\pi\rho_0LR^2}{3}$, so $E = \dfrac{\rho_0R^2}{3\epsilon_0s}$, the $1/s$ field of a line charge.

The two agree at $s = R$: the field is continuous because there is no surface charge.
:::
::::

```python
#| label: fig-gauss-profiles
#| caption: "Field magnitude versus distance for three Gauss's-law geometries (each scaled to 1 at the surface, R = 1): a uniform ball (linear inside, 1/r² outside), the cylinder of the example (s² inside, 1/s outside), and a uniform slab of half-thickness 1 (linear inside, constant outside)."
r = np.linspace(0, 3, 400)
fig, ax = plt.subplots()
ax.plot(r, np.where(r < 1, r, 1 / np.maximum(r, 1e-9)**2), lw=2, label="uniform ball")
ax.plot(r, np.where(r < 1, r**2, 1 / np.maximum(r, 1e-9)), lw=2, label=r"cylinder, $\rho\propto s$")
ax.plot(r, np.where(r < 1, r, 1.0), lw=2, label="uniform slab")
ax.axvline(1, color="gray", ls=":"); ax.set_xlabel("distance / R"); ax.set_ylabel(r"$E / E_{\rm surface}$"); ax.legend()
plt.show()
```

## Field Lines, and What Happens at a Surface Charge

Field lines start on positive charges and end on negative ones (or at infinity), and the number crossing a surface is proportional to the flux. Because $\nabla\times\vec{E} = \vec{0}$, they never close on themselves.

Applying Gauss's law to a thin pillbox straddling a surface with charge density $\sigma$, and the curl condition to a thin rectangular loop, gives the **boundary conditions**:

$$
E_{\perp}^{\text{above}} - E_{\perp}^{\text{below}} = \frac{\sigma}{\epsilon_0}, \qquad \vec{E}_{\parallel}^{\text{above}} = \vec{E}_{\parallel}^{\text{below}} .
$$ (eq-e-boundary)

The normal component jumps by $\sigma/\epsilon_0$; the tangential component is continuous. These are the starting point for every problem with conductors and dielectrics.

```python
#| label: fig-two-charges
#| caption: Field lines of +2q at x = −1 and −q at x = +1. Half the lines leaving the larger charge end on the smaller one and half escape to infinity. The field vanishes at the null point (black dot) on the axis beyond the negative charge.
x = np.linspace(-4, 7.5, 330); y = np.linspace(-4, 4, 240)
X, Y = np.meshgrid(x, y)
Ex = np.zeros_like(X); Ey = np.zeros_like(X)
for qq, x0 in [(2, -1), (-1, 1)]:
    d3 = ((X - x0)**2 + Y**2)**1.5 + 1e-6
    Ex += qq * (X - x0) / d3; Ey += qq * Y / d3
x_null = (np.sqrt(2) + 1) / (np.sqrt(2) - 1)        # solves 2/(x+1)^2 = 1/(x-1)^2 beyond the negative charge
fig, ax = plt.subplots(figsize=(7.5, 5.5))
ax.streamplot(X, Y, Ex, Ey, density=1.6, color="gray", linewidth=0.8)
ax.plot(-1, 0, "o", color="tab:red", ms=14); ax.plot(1, 0, "o", color="tab:blue", ms=10)
ax.plot(x_null, 0, "ko", ms=6)
ax.set_aspect("equal"); ax.set_xlim(-4, 7.5); ax.set_ylim(-4, 4); plt.show()
print(f"null point at x = {x_null:.3f}")
```

## Summary

- Coulomb's law plus superposition gives $\vec{E}$ of any charge distribution as the integral {eq}`eq-coulomb-integral`; use symmetry to kill components before integrating.
- Gauss's law, $\nabla\cdot\vec{E} = \rho/\epsilon_0$, holds always and solves for $\vec{E}$ when there is spherical, cylindrical or planar symmetry.
- Electrostatic fields are curl-free, so field lines never close.
- Across a surface charge, $E_\perp$ jumps by $\sigma/\epsilon_0$ and $\vec{E}_\parallel$ is continuous.
